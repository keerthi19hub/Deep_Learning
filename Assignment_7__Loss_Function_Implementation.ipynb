# Loss Function Implementation

## Objective

The objective of this assignment is to understand how neural networks measure prediction errors by implementing and comparing Mean Squared Error (MSE) and Cross-Entropy Loss. The assignment uses actual class labels and predicted outputs from a simple classification problem.


## Introduction

A loss function is used in machine learning to measure the difference between the actual values and the values predicted by a model. It helps the model understand how accurate or inaccurate its predictions are.

In this assignment, two loss functions are implemented:

1. Mean Squared Error (MSE)
2. Cross-Entropy Loss

The two loss functions are calculated using a simple binary classification example and their results are compared.


```python
import numpy as np
```

## Step 1: Define Actual and Predicted Values

For this classification problem, we use binary class labels where:

* 0 represents Class 0
* 1 represents Class 1

The actual labels are compared with the predicted probabilities produced by the model.


```python
# Actual class labels
actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
predicted = np.array([0.9, 0.2, 0.8, 0.6, 0.3])

print("Actual Labels:", actual)
print("Predicted Outputs:", predicted)
```

## Step 2: Calculate Mean Squared Error

Mean Squared Error (MSE) calculates the average squared difference between the actual values and predicted values.

The formula is:

MSE = (1/n) × Σ(Actual − Predicted)²

A lower MSE indicates that the predicted values are closer to the actual values.


```python
# Calculate Mean Squared Error
mse = np.mean((actual - predicted) ** 2)

print("Mean Squared Error:", mse)
```

## Step 3: Calculate Cross-Entropy Loss

Cross-Entropy Loss is commonly used for classification problems. It measures how well the predicted probabilities match the actual class labels.

For binary classification, the formula is:

Cross-Entropy Loss = −(1/n) × Σ[y log(p) + (1−y) log(1−p)]

A lower Cross-Entropy Loss indicates that the predicted probabilities are closer to the actual class labels.


```python
# Calculate Binary Cross-Entropy Loss

# Small value to avoid log(0)
epsilon = 1e-15

# Keep predicted probabilities between 0 and 1
predicted_clipped = np.clip(predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    actual * np.log(predicted_clipped) +
    (1 - actual) * np.log(1 - predicted_clipped)
)

print("Cross-Entropy Loss:", cross_entropy)
```

## Step 4: Compare the Results

The calculated values of Mean Squared Error and Cross-Entropy Loss are compared below.


```python
# Compare both loss functions

print("Loss Function Comparison")
print("------------------------")
print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

## Comparison Table

| Loss Function            | Calculated Value | Common Use              |
| ------------------------ | ---------------: | ----------------------- |
| Mean Squared Error (MSE) |            0.076 | Regression problems     |
| Cross-Entropy Loss       |            0.257 | Classification problems |

The numerical values of different loss functions should not be directly compared to decide which function is "better", because they use different formulas and scales. The important point is to understand what type of problem each loss function is designed for.


## Step 5: When Can Each Loss Function Be Used?

### Mean Squared Error (MSE)

Mean Squared Error is mainly used for regression problems where the output is a continuous numerical value.

For example, predicting:

* House prices
* Temperature
* Sales
* Stock values

MSE calculates the squared difference between the actual and predicted values.

### Cross-Entropy Loss

Cross-Entropy Loss is commonly used for classification problems where the model predicts classes or probabilities.

For example:

* Spam or not spam
* Disease or no disease
* Cat or dog
* Pass or fail

Cross-Entropy is useful because it works directly with predicted probabilities and gives a larger penalty when the model is confidently wrong.


## Step 6: Graphical Comparison

The following bar graph compares the Mean Squared Error and Cross-Entropy Loss obtained from the classification example.


```python
import matplotlib.pyplot as plt

# Loss function names
loss_names = ["Mean Squared Error", "Cross-Entropy Loss"]

# Loss values
loss_values = [mse, cross_entropy]

# Create bar graph
plt.figure(figsize=(8, 5))
plt.bar(loss_names, loss_values)

plt.title("Comparison of Loss Functions")
plt.xlabel("Loss Function")
plt.ylabel("Loss Value")

plt.show()
```
